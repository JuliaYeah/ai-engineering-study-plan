# Decision Trees; Random Forests; Gradient Boosting

## Learning Goal

Trees partition feature space; depth controls flexibility. Random forests average randomized trees to reduce variance. Boosting sequentially adds models to correct a loss-related residual signal.

## Minimal Prerequisites

Basic Python arrays and functions. Run all cells in order; no API key or network data is required. Data are synthetic teaching fixtures, not empirical market evidence.

## Guided Experiment

Increase tree depth and compare train/validation errors for tree, forest and boosting. Keep dataset and split fixed.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_friedman1
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
X,y=make_friedman1(n_samples=300,noise=2,random_state=7);rows=[]
for model in [DecisionTreeRegressor(max_depth=3,random_state=7),DecisionTreeRegressor(random_state=7),RandomForestRegressor(n_estimators=30,max_depth=5,random_state=7),GradientBoostingRegressor(n_estimators=50,max_depth=2,random_state=7)]:
    model.fit(X[:200],y[:200]); rows.append([str(model).split("(")[0],mean_absolute_error(y[:200],model.predict(X[:200])),mean_absolute_error(y[200:],model.predict(X[200:]))])
result=pd.DataFrame(rows,columns=["model","train","validation"]);print(result);result.set_index("model").plot.bar();plt.tight_layout();plt.show()
```

## Expected Observations

A deep tree can achieve tiny training error with poor generalization. The best model on this fixture is not proof of superiority on market data.
