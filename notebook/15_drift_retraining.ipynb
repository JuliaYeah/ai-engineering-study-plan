# Drift; Retraining

## Learning Goal

Input drift is a distribution change; performance drift requires outcomes. A drift alarm should trigger diagnosis before retraining, since retraining on bad labels can make the system worse.

## Minimal Prerequisites

## Guided Experiment

Change the shift size and compare input histograms and mean residuals. Describe how delayed five-day labels postpone performance monitoring.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rng=np.random.default_rng(7);reference=rng.normal(size=300);current=rng.normal(1,1,300)
print("standardized mean shift",(current.mean()-reference.mean())/reference.std())
plt.hist(reference,bins=20,alpha=.5,label="reference");plt.hist(current,bins=20,alpha=.5,label="current");plt.legend();plt.show()
print("Available performance labels at origin 100 end before 100; five-day labels from recent origins are pending.")
```

## Expected Observations
