# Supervised and Unsupervised Learning; Baselines

## Learning Goal

Supervised learning uses a specified target; clustering groups features without a target. A simple baseline makes a model score interpretable. Predicting the training mean is a valid regression baseline, but it is not a competitive financial-volatility baseline.

## Minimal Prerequisites

Basic Python arrays and functions. Run all cells in order; no API key or network data is required. Data are synthetic teaching fixtures, not empirical market evidence.

## Guided Experiment

Increase the noise level and compare Ridge with a training-mean predictor. Change the clustering seed and compare group labels with the continuous regression target.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge
from sklearn.cluster import KMeans
from sklearn.metrics import mean_absolute_error
X,y=make_regression(n_samples=300,n_features=4,noise=20,random_state=7)
for model in [DummyRegressor(),Ridge()]:
    model.fit(X[:200],y[:200]); print(type(model).__name__,mean_absolute_error(y[200:],model.predict(X[200:])))
labels=KMeans(n_clusters=3,random_state=7,n_init=10).fit_predict(X)
plt.scatter(X[:,0],X[:,1],c=labels); plt.title("Feature clusters, not target classes"); plt.show()
```

## Expected Observations

Ridge should beat the mean when the target depends on the features. Clusters do not automatically correspond to useful prediction classes.
