# Feature Engineering; Interpretability; Feature Importance

## Learning Goal

Features encode information available at decision time. Permutation importance measures the loss in held-out performance after disrupting a feature; correlated features can substitute for each other and dilute importance.

## Minimal Prerequisites

## Guided Experiment

Duplicate the strongest feature and compare permutation importance before and after. Explain why feature importance is not a causal effect.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge
from sklearn.inspection import permutation_importance
X,y=make_regression(n_samples=300,n_features=4,n_informative=2,noise=5,random_state=7);X=np.column_stack([X,X[:,0]])
model=Ridge().fit(X[:200],y[:200]);result=permutation_importance(model,X[200:],y[200:],n_repeats=10,random_state=7)
print(result.importances_mean);plt.bar(range(5),result.importances_mean);plt.xlabel("Feature (4 duplicates 0)");plt.show()
```

## Expected Observations

Correlated features make attribution ambiguous. Importance is conditional on model, dataset and metric, not an intrinsic causal ranking.
